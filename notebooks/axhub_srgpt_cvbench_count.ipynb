```python
import os
import sys
from pathlib import Path

# 프로젝트 루트 (axhub에서 경로 수정)
PROJECT_ROOT = Path(os.getcwd()).resolve()
if (PROJECT_ROOT / "test_fixed_specialist_mas_v2.py").exists():
    pass  # 이미 루트
elif (PROJECT_ROOT.parent / "test_fixed_specialist_mas_v2.py").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
else:
    PROJECT_ROOT = Path("/home/jovyan/Spatial_MAS")  # axhub 기본, 없으면 수정

os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))
os.environ["SPATIALRGPT_PATH"] = str(PROJECT_ROOT / "SpatialRGPT")

print("PROJECT_ROOT:", PROJECT_ROOT)
print("SPATIALRGPT_PATH:", os.environ.get("SPATIALRGPT_PATH"))
```

```python
from test_fixed_specialist_mas_v2 import build_runners_fixed, run_fixed_specialist_mas_test

# 모델 로드 (처음 1회, 수 분 소요)
head_gen, spec_gen, reason_gen = build_runners_fixed(
    specialist="spatial_rgpt",
    specialist_device="cuda",
    use_vlm_reasoning=True,
)
print("모델 로드 완료")
```

```python
# 빠른 확인: 10개
results = run_fixed_specialist_mas_test(
    head_gen, spec_gen, reason_gen,
    specialist="spatial_rgpt",
    benchmark="cvbench",
    max_samples=10,
    category_filter=["Count"],
)

print(f"\nAccuracy: {results['correct']}/{results['total']} = {100*results['accuracy']:.1f}%")
```

```python
# 전체 100개 평가 (위 셀 통과 후 실행)
results_full = run_fixed_specialist_mas_test(
    head_gen, spec_gen, reason_gen,
    specialist="spatial_rgpt",
    benchmark="cvbench",
    max_samples=100,
    category_filter=["Count"],
)

print(f"\n최종 Accuracy: {results_full['correct']}/{results_full['total']} = {100*results_full['accuracy']:.1f}%")
```
